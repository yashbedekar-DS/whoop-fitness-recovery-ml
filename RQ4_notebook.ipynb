# RQ4: Feature Importance and Interpretability

Research question: Which features contribute most to recovery score prediction?

Dataset: WHOOP Fitness Dataset. Target variable: `recovery_score`.

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import Image, display


def find_project_root() -> Path:
    """Locate the assignment root regardless of the notebook working directory."""
    cwd = Path.cwd().resolve()
    candidates = [cwd, *cwd.parents]

    assignment_dir = cwd / 'Machine_Learning_Assignment'
    if assignment_dir.exists():
        candidates.insert(1, assignment_dir.resolve())

    for candidate in candidates:
        if (candidate / 'whoop_assignment_pipeline.py').exists():
            return candidate

    raise FileNotFoundError(
        'Could not locate Machine_Learning_Assignment/whoop_assignment_pipeline.py from the current working directory.'
    )


ROOT = find_project_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from whoop_assignment_pipeline import DATA_PATH, run_rq4

table_path = ROOT / 'outputs' / 'tables' / 'RQ4_aggregated_feature_importance.csv'
image_dir = ROOT / 'outputs' / 'figures' / 'images'
image_names = ['RQ4_feature_importance_bar.png', 'RQ4_correlation_heatmap.png']

df = pd.read_csv(DATA_PATH)
print(f"Project root: {ROOT}")
print(f"Dataset shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
display(df.head())

print('This notebook displays the executed RQ4 outputs generated by whoop_assignment_pipeline.py.')

if not table_path.exists() or any(not (image_dir / image_name).exists() for image_name in image_names):
    print('RQ4 outputs were missing, so they are being regenerated now...')
    run_rq4()

if table_path.exists():
    display(pd.read_csv(table_path).head(15))
else:
    print(f'Could not find the RQ4 feature importance table at: {table_path}')

for image_name in image_names:
    image_path = image_dir / image_name
    if image_path.exists():
        display(Image(filename=str(image_path)))
    else:
        print(f'Missing RQ4 image: {image_path}')
```
